```python
import pandas as pd

df1 = pd.read_csv("DataSetLimpio.csv")
df2 = pd.read_csv("DataSetLimpio_V2.csv")

print("="*70)
print("COMPARACIÓN DE DATASETS")
print("="*70)

# Dimensiones
print("\nDimensiones")
print("-----------------------------")
print("DataSetLimpio.csv        :", df1.shape)
print("DataSetLimpio_V2   :", df2.shape)

# Columnas
print("\nColumnas")
print("-----------------------------")

cols1 = set(df1.columns)
cols2 = set(df2.columns)

print("Columnas iguales:", cols1 == cols2)

print("Solo en DataSetLimpio:")
print(cols1 - cols2)

print("Solo en DataSetLimpio_V2:")
print(cols2 - cols1)

# Tipos de datos
print("\nTipos de datos")
print("-----------------------------")

display(pd.DataFrame({
    "DataSetLimpio": df1.dtypes,
    "DataSetLimpio_V2": df2.dtypes
}))

# Nulos
print("\nValores nulos")
print("-----------------------------")

display(pd.DataFrame({
    "DataSetLimpio": df1.isna().sum(),
    "DataSetLimpio_V2": df2.isna().sum()
}))

# Duplicados
print("\nDuplicados")
print("-----------------------------")

print("DataSetLimpio :", df1.duplicated().sum())
print("DataSetLimpio_V2 :", df2.duplicated().sum())
```
